# Player Energy Spike Analysis: Sliding Window Technique
**Objective:** Find the maximum average energy boost within a fixed time window $k$ from a stream of player energy readings.


```python
import time
import numpy as np
import matplotlib.pyplot as plt

# 1. Naive Approach: O(n * k)
def max_average_naive(nums, k):
    max_sum = float('-inf')
    # Loop through all possible starting positions
    for i in range(len(nums) - k + 1):
        current_sum = sum(nums[i:i+k])
        if current_sum > max_sum:
            max_sum = current_sum
    return max_sum / k

# 2. Optimized Sliding Window Approach: O(n)
def max_average_sliding_window(nums, k):
    # Calculate sum of the first window
    current_window_sum = sum(nums[:k])
    max_sum = current_window_sum
    
    # Slide the window across the array
    for i in range(k, len(nums)):
        # Add the next element, subtract the element leaving the window
        current_window_sum += nums[i] - nums[i - k]
        if current_window_sum > max_sum:
            max_sum = current_window_sum
            
    return max_sum / k

# Test Case
energy_readings = [1, 12, -5, -6, 50, 3]
window_size = 4

print(f"Naive Result: {max_average_naive(energy_readings, window_size)}")
print(f"Sliding Window Result: {max_average_sliding_window(energy_readings, window_size)}")
```

```python
# Generate synthetic data: 10,000 tournament energy readings
np.random.seed(42)
large_dataset = np.random.randint(-100, 100, size=10000).tolist()
k_size = 500

# Benchmark Naive
start = time.time()
max_average_naive(large_dataset, k_size)
time_naive = time.time() - start

# Benchmark Sliding Window
start = time.time()
max_average_sliding_window(large_dataset, k_size)
time_window = time.time() - start

print(f"Naive Time for 10k items: {time_naive:.5f} seconds")
print(f"Sliding Window Time for 10k items: {time_window:.5f} seconds")

# Plot the comparison
plt.figure(figsize=(6, 4))
plt.bar(['Naive O(N*K)', 'Sliding Window O(N)'], [time_naive, time_window], color=['tomato', 'mediumseagreen'])
plt.ylabel('Time (seconds)')
plt.title('Execution Time Comparison (10,000 Elements)')
plt.show()
```
